**Visualize** คือการสรุปผลให้คนอื่นเข้าใจง่ายๆ

```
import pandas as pd
```

```
df = pd.read_csv('https://raw.github.com/pandas-dev/pandas/master/pandas/tests/io/data/csv/iris.csv')
df
```

# Parallel Coordinates

```
pd.plotting.parallel_coordinates(df, 'Name');
```

## ปรับ parameter: color

### HTML colors

FF6600

336600

ECFF00

เลขพวกนี้ คือ รหัสสี เรียกว่า **HTML colors** ต้อวมี **#** ข้างหน้าเสมอ

```
pd.plotting.parallel_coordinates(df, 'Name', color=('#FF6600', '#336600', '#ECFF00')) #html color
```

### ใส่ตัวย่อภาษาอังกฤษของสี

```
pd.plotting.parallel_coordinates(df, 'Name', color=('r', 'g', 'b')) #ตัวย่อของสี R G B W C M Y K
```

### **rgb value** (r,g,b) | r,g,b -> [0,1]
เราสามารถกำหนดปริมาณของแสงที่จะมาผสมกันได้ ซึ่ง r,g,b จะมีค่าตั้งแต่ 0 ถึง 1 เท่านั้น

```
pd.plotting.parallel_coordinates(df, 'Name', color=((1,0,0.5), (0,0.7,0.3), (0.1,0,0.5))) #rgb value (r,g,b) | r,g,b -> [0,1]
```

## ปรับ parameter: colormap
การที่เราจะมาเลือกแต่ละสีเพื่อโชว์ แต่ **colormap เป็นชุดสีที่เขาเลือกมาให้ละ มันจะได้เป็นสีเป็นโทนเดียวกัน** มันจะเข้ากันมากกว่า

```
pd.plotting.parallel_coordinates(df, 'Name', colormap='cool')
```

```
pd.plotting.parallel_coordinates(df, 'Name', colormap='magma')
```

# เริ่ม Matplotlib

```
from matplotlib import pyplot as plt
```

### แปลง data จาก pandas dataframe เป็น numpy array

```
np_data = df.iloc[:,:-1].to_numpy() #เอาทุกแถว และ col. แรกจนถึงก่อนสุดท้าย
np_data
```

# Visualize array data ด้วย pixel
เราจะแทนค่าข้างในแต่ละค่าด้วยจุดสี

pixel คือ หน่วยเล็กสุดของภาพ

**interpolation='nearest'** การบอกว่าทำให้มันแยกกันชัดๆ

```
import matplotlib
```

```
matplotlib.rcParams['figure.figsize']=[15,15] #กำหนดขนาดของ figure กำหนดว่ารูปของเราจะให้มันเล็กมันใหญ่แค่ไหน

plt.imshow(np_data[:50,:],interpolation='nearest')
```

```
plt.imshow(np_data[50:100,:],interpolation='nearest')
```

# การใช้ subplot
การเอากราฟมาเทียบกัน

```
plt.subplot(1,3,1)  # subplot(มีกี่แถว, มีกี่หลัก, เป็นตัวที่เท่าไหร่)
plt.imshow(np_data[:50,:],interpolation='nearest',cmap='magma')

plt.subplot(1,3,2)
plt.imshow(np_data[50:100,:],interpolation='nearest',cmap='magma')

plt.subplot(1,3,3)
plt.imshow(np_data[100:150,:],interpolation='nearest',cmap='magma')
```
